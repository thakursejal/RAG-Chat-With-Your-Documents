```python
# ============================================
# 2️⃣ EMBEDDINGS & SEMANTIC SIMILARITY
# ============================================

from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

# Load the embedding model
embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

print("✅ Embedding model loaded successfully!")

# Sample documents
documents = [
    "Machine learning allows computers to learn patterns from data.",
    "Artificial intelligence enables machines to perform intelligent tasks.",
    "Python is a popular programming language for data science.",
    "Neural networks are inspired by the structure of the human brain."
]

# Generate embeddings
document_embeddings = embedding_model.encode(documents)

print("\n📄 Number of documents:", len(documents))
print("📐 Embedding dimension:", document_embeddings.shape[1])

# User query
query = "How do computers learn from data?"

# Generate query embedding
query_embedding = embedding_model.encode([query])

# Calculate similarity
similarities = cosine_similarity(
    query_embedding,
    document_embeddings
)[0]

print("\n🔎 Query:", query)
print("\n📊 Semantic Similarity Scores:")

for i, score in enumerate(similarities):
    print(f"\nDocument {i + 1}: {score:.4f}")
    print(documents[i])

# Find the most relevant document
best_index = similarities.argmax()

print("\n⭐ Most Relevant Document:")
print(documents[best_index])
print("Similarity Score:", round(similarities[best_index], 4))
```

```python
# ============================================
# 3️⃣ TEXT CHUNKING + VECTOR DATABASE
# ============================================

from sentence_transformers import SentenceTransformer
import chromadb

# Load embedding model
embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

print("✅ Embedding model loaded!")

# Sample knowledge base created for this project
knowledge_text = """
Artificial Intelligence is a field of computer science that enables
machines to perform tasks that normally require human intelligence.

Machine Learning is a subset of Artificial Intelligence. It allows
computers to learn patterns from data and make predictions or decisions
without being explicitly programmed for every task.

Deep Learning is a subset of Machine Learning that uses neural networks
with multiple layers to learn complex patterns from large amounts of data.

Natural Language Processing, or NLP, enables computers to understand,
process, and generate human language.

Generative AI refers to AI systems that can generate new content such as
text, images, audio, and code.

Retrieval-Augmented Generation, or RAG, combines information retrieval
with language generation. A RAG system retrieves relevant information
from a knowledge base and provides it to an LLM as context before
generating an answer.

Embeddings are numerical representations of text. Similar pieces of text
usually have similar embedding representations.

A vector database stores vector representations such as embeddings and
allows applications to efficiently search for similar information.
"""

# --------------------------------------------
# TEXT CHUNKING
# --------------------------------------------

def create_chunks(text, chunk_size=60):
    words = text.split()
    chunks = []

    for i in range(0, len(words), chunk_size):
        chunks.append(" ".join(words[i:i + chunk_size]))

    return chunks


chunks = create_chunks(knowledge_text)

print("\n✅ Text chunking completed!")
print("Number of chunks:", len(chunks))

for i, chunk in enumerate(chunks):
    print(f"\n--- Chunk {i + 1} ---")
    print(chunk)


# --------------------------------------------
# CREATE EMBEDDINGS
# --------------------------------------------

chunk_embeddings = embedding_model.encode(chunks)

print("\n✅ Chunk embeddings created!")
print("Embedding dimensions:", chunk_embeddings.shape)


# --------------------------------------------
# CREATE CHROMADB VECTOR DATABASE
# --------------------------------------------

chroma_client = chromadb.Client()

collection = chroma_client.get_or_create_collection(
    name="codomax_rag_knowledge"
)

# Add chunks and embeddings
collection.add(
    ids=[f"chunk_{i}" for i in range(len(chunks))],
    documents=chunks,
    embeddings=chunk_embeddings.tolist()
)

print("\n✅ ChromaDB vector database created!")
print("Documents stored:", collection.count())
```

```python
# ============================================
# 4️⃣ SEMANTIC RETRIEVAL FROM CHROMADB
# ============================================

query = "What is Retrieval-Augmented Generation?"

# Convert the question into an embedding
query_embedding = embedding_model.encode([query])

# Search ChromaDB for the most relevant chunks
results = collection.query(
    query_embeddings=query_embedding.tolist(),
    n_results=2
)

print("🔎 User Question:")
print(query)

print("\n📚 Retrieved Context:")
print("=" * 60)

for i, document in enumerate(results["documents"][0]):
    print(f"\n--- Retrieved Chunk {i + 1} ---")
    print(document)

print("\n✅ Semantic retrieval completed!")
```

```python
from google.colab import userdata
from huggingface_hub import InferenceClient

HF_TOKEN = userdata.get("HF_TOKEN")

client = InferenceClient(api_key=HF_TOKEN)

print("✅ Hugging Face token loaded securely!")
```

```python
# ============================================
# STEP 4: CONNECT RETRIEVED CONTEXT TO LLM
# ============================================

from huggingface_hub import InferenceClient

MODEL = "openai/gpt-oss-120b:fastest"

def ask_rag(question):

    # 1. Convert question into an embedding
    question_embedding = embedding_model.encode([question])

    # 2. Search the vector database
    results = collection.query(
        query_embeddings=question_embedding.tolist(),
        n_results=2
    )

    # 3. Get the retrieved documents
    retrieved_documents = results["documents"][0]

    # 4. Combine retrieved documents into context
    context = "\n\n".join(retrieved_documents)

    # 5. Create a grounded prompt
    prompt = f"""
You are a helpful AI assistant.

Answer the question using ONLY the information
provided in the context below.

If the answer is not available in the context,
say:
"The information is not available in the knowledge base."

Context:
{context}

Question:
{question}

Answer:
"""

    # 6. Send context + question to the LLM
    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],
        max_tokens=200
    )

    answer = response.choices[0].message.content

    return answer, context


# Test the RAG system
question = "What is Retrieval-Augmented Generation?"

answer, context = ask_rag(question)

print("🔎 QUESTION")
print(question)

print("\n📚 RETRIEVED CONTEXT")
print("=" * 60)
print(context)

print("\n🤖 RAG ANSWER")
print("=" * 60)
print(answer)

print("\n✅ RAG pipeline completed successfully!")
```

# 2️⃣ Embeddings & Semantic Similarity

## What are Embeddings?

Embeddings are numerical representations of text.

They convert words, sentences, or documents into vectors
(numbers) that capture their meaning and relationships.

For example:

- "Machine learning uses data"
- "Computers learn patterns from data"

These sentences have different words but similar meanings,
so their embeddings can be close to each other.

---

## What is Semantic Similarity?

Semantic similarity measures how similar two pieces of text
are based on their meaning rather than only matching exact words.

A common method is **Cosine Similarity**.

### Cosine Similarity

Cosine similarity compares two vectors and produces a score
that represents how similar they are.

A higher score generally indicates greater semantic similarity.

---

## Example

User Question:

**"How do computers learn from data?"**

The system converts the question into an embedding and
compares it with document embeddings.

The document with the highest similarity score is considered
more relevant to the question.

---

## Embedding Model Used

In this project, the following Sentence Transformers model
is used:

**all-MiniLM-L6-v2**

It converts text into **384-dimensional vectors**.

---

## Why Embeddings are Important in RAG

Embeddings allow a RAG system to:

1. Convert documents into vectors
2. Convert the user's question into a vector
3. Compare their meanings
4. Find relevant information
5. Retrieve useful context for the LLM

---

## Key Takeaway

**Text → Embedding → Vector → Similarity Search → Relevant Information**

# 3️⃣ Document Processing & Text Chunking

## What is Document Processing?

Document processing is the process of extracting useful text
from documents so that an AI system can understand and retrieve
the required information.

Common document sources include:

- PDF files
- Text files
- Word documents
- Web pages
- Reports
- Knowledge bases

---

## What is Text Chunking?

Large documents are divided into smaller pieces called **chunks**.

Instead of sending an entire document to an LLM, the RAG system
retrieves only the relevant chunks needed to answer a question.

---

## Why is Chunking Important?

Chunking helps to:

- Improve retrieval accuracy
- Reduce unnecessary information
- Manage large documents
- Reduce the amount of context sent to the LLM
- Improve response relevance

---

## Chunking Process

**Document → Extract Text → Split into Chunks → Create Embeddings**

Each chunk is then converted into an embedding and stored
in the vector database.

---

## Chunking Used in This Project

The project uses a simple word-based chunking approach.

The knowledge text is divided into chunks of approximately
60 words each.

These chunks are then converted into embeddings and stored
in ChromaDB.

---

## Key Takeaway

**Large Document → Smaller Chunks → Embeddings → Vector Database**

Chunking is an important step because good chunks help the
retrieval system find the most relevant information.

# 4️⃣ Vector Databases & Similarity Search

## What is a Vector Database?

A vector database is a database designed to store and search
numerical vector representations called embeddings.

In a RAG system, document chunks are converted into embeddings
and stored in a vector database.

---

## ChromaDB

This project uses **ChromaDB** as the vector database.

ChromaDB stores:

- Document chunks
- Embeddings
- Unique IDs
- Metadata when required

---

## How Similarity Search Works

When a user asks a question:

1. The question is converted into an embedding.
2. The vector database compares the question embedding with
   stored document embeddings.
3. The most similar chunks are retrieved.
4. The retrieved chunks are provided to the LLM as context.
5. The LLM generates a grounded answer.

---

## RAG Retrieval Pipeline

| Step | Process |
|---|---|
| 1️⃣ | 👤 **User Question** |
| ↓ | |
| 2️⃣ | 🧠 **Question Embedding** |
| ↓ | |
| 3️⃣ | 🗄️ **Vector Database Search** |
| ↓ | |
| 4️⃣ | 📚 **Relevant Chunks** |
| ↓ | |
| 5️⃣ | 📝 **Context** |
| ↓ | |
| 6️⃣ | 🤖 **LLM** |
| ↓ | |
| 7️⃣ | 💡 **Final Answer** |

### Key Idea

The RAG system first finds relevant information from the
knowledge base and then gives that information to the LLM
so it can generate a grounded answer.

## Benefits of Vector Databases

- Fast similarity search
- Efficient handling of large collections
- Semantic rather than exact keyword matching
- Useful for document-based AI applications
- Supports scalable RAG systems

---

## Key Takeaway

A vector database acts as the searchable knowledge layer
of a RAG application.

It helps the LLM receive relevant information before generating
the final response.
